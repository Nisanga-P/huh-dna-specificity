```python
import torch
import numpy as np
```

```python
embedding_tensor = torch.load("../data/embeddings/full_sDBM_embeddings.pt").numpy()
x_input_vector = torch.load("../data/embeddings/X_input_vectors.pt").numpy()
y_target_labels = torch.load("../data/embeddings/y_target_labels.pt").numpy()
```

```python
X_train = torch.load("../data/embeddings/ori_predictor_combined/combined_X_train.pt")
y_train = torch.load("../data/embeddings/ori_predictor_combined/combined_y_train.pt")
X_val_rep = torch.load("../data/embeddings/ori_predictor_combined/X_val_rep.pt")
y_val_rep = torch.load("../data/embeddings/ori_predictor_combined/y_val_rep.pt")
X_val_relax = torch.load("../data/embeddings/ori_predictor_combined/X_val_relax.pt")
y_val_relax = torch.load("../data/embeddings/ori_predictor_combined/y_val_relax.pt")

X_test = torch.concat([X_val_rep, X_val_relax], dim=0)
y_test = torch.concat([y_val_rep, y_val_relax], dim=0)
```

```python
y_target_labels.shape
```

```python
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification
```

```python
X_train,  X_test, y_train, y_test = train_test_split(x_input_vector, y_target_labels, random_state=3242, test_size=0.2)

print(X_train.shape)

y_train.shape
```

```python
model = SVC(C=7500)

model.fit(X_train, y_train)

model.score(X_test, y_test)
```

```python
# scaling makes a massive difference

clf = make_pipeline(StandardScaler(), SVC(kernel='rbf', C=5, gamma='scale'))

clf.fit(X_train, y_train)

accuracy = clf.score(X_test, y_test)

print(accuracy)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
model = RandomForestClassifier(
    n_estimators=1000, 
    max_depth=None, 
    random_state=42
)


model.fit(X_train, y_train)


y_pred = model.predict(X_test)         # Hard labels: 0 or 1
y_probs = model.predict_proba(X_test)  # Class probabilities [prob_0, prob_1]


print(f"Accuracy Score: {accuracy_score(y_test, y_pred):.4f}\n")
print("Classification Report:")
print(classification_report(y_test, y_pred))
```
