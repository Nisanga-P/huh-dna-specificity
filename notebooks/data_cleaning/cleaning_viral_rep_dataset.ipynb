```python
import pandas as pd
import numpy as np
import random
```

```python
original_df_rep = pd.read_csv("../data/raw/positive_rep_virus_dataset.csv")

original_df_rep
```

```python
# Remove NaN ori
original_df_rep.dropna(
    subset=['ori_seq', 'sDBM_pos', 'rep_amino_acid_seq'],
    inplace=True
)

# Drop entries where sDBM confidence < 10
df_rep_cleaned = original_df_rep[original_df_rep['sDBM_sim_gap'] >= 10]

# Drop entries where ori score < 130
df_rep = df_rep_cleaned[df_rep_cleaned['ori_align_score'] >= 130]

# discard unnecessary columns
df_rep = df_rep.drop(columns=['taxid', 'virus_name', 'sDBM_sim_gap', 'ori_align_score', 'iteration_name'])

# only keep unique sDBM-ori pairs
df_rep = df_rep.drop_duplicates(subset=["sDBM", "ori_seq"], keep='first')

len(df_rep)
```

```python
df_rep.to_csv('../data/processed/cleaned_datasets/unique_rep_oriR_dataset.csv', index=False)
```
