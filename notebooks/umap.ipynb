```python
import matplotlib.pyplot as plt
from umap import UMAP
from sklearn.preprocessing import StandardScaler

import seaborn as sns

import torch
```

```python
x_input_vector = torch.load("../data/embeddings/X_input_vectors.pt").numpy()
y_target_labels = torch.load("../data/embeddings/y_target_labels.pt").numpy()
```

```python

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(x_input_vector)

# Reduce dimensions
reducer = UMAP(n_neighbors=8, min_dist=0.5, random_state=42)
embedding = reducer.fit_transform(X_scaled)

# Visualize
plt.scatter(embedding[:, 0], embedding[:, 1], c=y_target_labels, cmap='coolwarm', s=5)
plt.title("UMAP Binary Classification Visualization")
plt.show()
```

```python
sDBM_input_vector = torch.load("../data/embeddings/sDBM_vectors.pt").numpy()
ori_target_labels = torch.load("../data/embeddings/ori_class_labels.pt").numpy()

pos_1_labels = ori_target_labels[:, 0]
pos_2_labels = ori_target_labels[:, 1]
pos_3_labels = ori_target_labels[:, 2]
pos_4_labels = ori_target_labels[:, 3]
pos_5_labels = ori_target_labels[:, 4]
pos_6_labels = ori_target_labels[:, 5]
pos_7_labels = ori_target_labels[:, 6]
pos_8_labels = ori_target_labels[:, 7]
pos_9_labels = ori_target_labels[:, 8]
```

```python
def umap_ori_pos(pos_labels):
    # Scale features
    scaler = StandardScaler()
    X_scaled2 = scaler.fit_transform(sDBM_input_vector)

    # Reduce dimensions
    reducer = UMAP(n_neighbors=5, min_dist=0.5, random_state=42)
    embedding = reducer.fit_transform(X_scaled2)

    # Visualize
    plt.figure(figsize=(8, 6))
    sns.scatterplot(
        x=embedding[:, 0],
        y=embedding[:, 1],
        hue=pos_labels,
        palette="Set1",
        legend="full",
    )
    plt.title("UMAP Multi-Class Visualization")
    plt.show()
```
